```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
import dagshub
import mlflow

```

```python
url = 'https://raw.githubusercontent.com/campusx-official/toy-datasets/main/student_performance.csv'
df = pd.read_csv(url)
df.head()
```

```python
X = df.drop(columns=['Placed'])
y = df['Placed']

# data spliting
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state= 42, test_size=0.2)
```

```python
import dagshub
dagshub.init(repo_owner='akashkangule18', repo_name='ml-ops-project', mlflow=True)
```

```python

mlflow.set_experiment("exp-1-baseline-model-1")

with mlflow.start_run():
    mlflow.log_param('test_size ', 0.2)
    mlflow.log_param('random state', 42)

    # applying linear regression
    from sklearn.linear_model import LogisticRegression

    lr = LogisticRegression()

    lr.fit(X_train,y_train)

    y_pred = lr.predict(X_test)


    from sklearn.metrics import accuracy_score, recall_score, precision_score, confusion_matrix

    acc = accuracy_score(y_test,y_pred)
    rcc = recall_score(y_test,y_pred)
    pcc = precision_score(y_test,y_pred)

    mlflow.log_metric('accuracy_score',acc)
    mlflow.log_metric('recall_score', rcc)
    mlflow.log_metric('precidion_score', pcc)

    mlflow.sklearn.log_model(lr,"LogisticRegression")

    # log notebook
    mlflow.log_artifact("exp-1.ipynb")


    print('accuracy_score: ', acc)
    print('recall: ', rcc)
    print('precison', pcc)
```
