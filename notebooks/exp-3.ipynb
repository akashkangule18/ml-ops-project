```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
import dagshub
import mlflow
import xgboost
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
url = 'https://raw.githubusercontent.com/campusx-official/toy-datasets/main/student_performance.csv'
df = pd.read_csv(url)
df.head()
```

```python
X = df.drop(columns=['Placed'])
y = df['Placed']

# data spliting
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state= 42, test_size=0.2)
```

```python
import dagshub
dagshub.init(repo_owner='akashkangule18', repo_name='ml-ops-project', mlflow=True)
```

```python
mlflow.set_experiment("exp-4 with final model")

with mlflow.start_run(run_name='final_model'):
    lr = RandomForestClassifier(
        n_estimators= 50,
        max_depth=2,
        min_impurity_decrease=0.01
    )
    print('RandomForestClassifier')

    lr.fit(X_train,y_train)
    lr.predict(X_test)

    y_pred = lr.predict(X_test)

    from sklearn.metrics import accuracy_score, recall_score, precision_score, confusion_matrix

    acc = accuracy_score(y_test,y_pred)
    rcc = recall_score(y_test,y_pred)
    pcc = precision_score(y_test,y_pred)

    # log params
    mlflow.log_param('test_size' , 0.2)
    mlflow.log_params(lr.get_params())

    # log code file
    mlflow.log_artifact('exp-3.ipynb')

    # signature
    signature = mlflow.models.infer_signature(X_train, lr.predict(X_train))

    # log model 
    mlflow.sklearn.log_model(
        sk_model= lr,
        name = 'RandomForestClassifier',
        signature= signature
    )


    print('RandomForestClassifier')
    print('accuracy_score', acc)
    print('recall_score', rcc)
    print('precision score', pcc)
```
